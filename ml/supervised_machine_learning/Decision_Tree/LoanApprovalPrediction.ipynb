```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder

data = {
    "income": [
        25000, 30000, 35000, 45000, 50000,
        60000, 70000, 80000, 28000, 40000
    ],

    "credit_score": [
        580, 600, 620, 680, 700,
        720, 750, 780, 590, 650
    ],

    "loan_amount": [
        500000, 400000, 300000, 300000, 400000,
        500000, 600000, 700000, 250000, 250000
    ],

    "employment": [
        "No", "Yes", "Yes", "Yes", "Yes",
        "Yes", "Yes", "Yes", "No", "Yes"
    ],

    "decision": [
        "Reject", "Reject", "Reject", "Approve", "Approve",
        "Approve", "Approve", "Approve", "Reject", "Approve"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["employment"] = df["employment"].map({
    "No": 0,
    "Yes": 1
})

df["decision"] = df["decision"].map({
    "Reject": 0,
    "Approve": 1
})
```

```python
X = df[
    [
        "income",
        "credit_score",
        "loan_amount",
        "employment"
    ]
]

y = df["decision"]
```

```python
model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=3,
    random_state=42
)

model.fit(X, y)
```

```python
plt.figure(figsize=(16, 9))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=["Reject", "Approve"],
    filled=True,
    rounded=True
)

plt.show()
```
