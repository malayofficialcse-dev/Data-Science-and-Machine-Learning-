```python
import pandas as pd

data = {
    "age": [20, 25, 30, 35, 40],
    "salary": [20000, 35000, 50000, 70000, 100000]
}

df= pd.DataFrame(data)

print(df)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler() 
scaled_data=scaler.fit_transform(df)
print(scaled_data)
```
